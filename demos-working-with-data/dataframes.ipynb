# NumPy & Pandas

```python
import numpy as np 
import pandas as pd
```

## NumPy

**Numpy Arrays**

```python
int_array = np.array([1,2,3,4,5])
print(int_array)
float_array = np.array([3.14, 2.71, 1.61, 1.41, 1.21])
print(float_array)
string_array = np.array(['a', 'b', 'c', 'd', 'e'])
print(string_array)

# create a numpy array using arange
range_array = np.arange(1, 6)
print(range_array)

# create a numpy array using randint
randint_array = np.random.randint(1, 10, size=(5))
print(randint_array)

# info about an array
print(f'Size: {string_array.size}')

# indexes
print(f'First element: {float_array[0]}')
print(f'Last element: {float_array[-1]}')

# update an array
int_array[0] = 10
print(int_array)

# slicing
print(f'First three elements: {float_array[:3]}')

# copy an array
array_copy = float_array.copy()
print(array_copy)

# array concatenation
array_concat = np.concatenate((int_array, [6,7,8,9,10]))
print(array_concat)

print(string_array.size)
```

**using arithmetic operators with numpy arrays**

```python
x = np.arange(10)
print(x)

# add 5 to each element
print(x + 5)

# multipy each element by 10
print(x * 10)
```

**Numpy array functions**

```python
print(np.power(x, 2))
print(np.max(x))
print(np.min(x))
print(np.mean(x))
```

**Array sorting**

```python
# sort an array as a copy
print(np.sort(float_array))

# sort an array in place
float_array.sort()
print(float_array)
```

## Pandas

**Series**

```python
# pandas series
series = pd.Series([1,2,3,4,5])
print(series)

# pandas series with custom index, built from a dictionary
spanish_nums = {'one': 'uno', 'two': 'dos', 'three': 'tres', 
                'four': 'cuatro', 'five': 'cinco'}
spanish_nums_series = pd.Series(spanish_nums)
print(spanish_nums_series)

```

**DataFrames**

```python
states_areas = pd.Series({'California': 423967, 'Texas': 695662, 'New York': 141297,
             'Florida': 170312, 'Illinois': 149995})
states_pop = pd.Series({'California': 38332521, 'Texas': 26448193, 
                        'New York': 19651127, 'Illinois': 12882135, 'Florida': 19552860})

states_df = pd.DataFrame({'area': states_areas, 
                          'population': states_pop})
states_df
```
