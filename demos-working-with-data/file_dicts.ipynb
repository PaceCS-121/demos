**Reading Lines**

```python
# open the file
olympics = open('data/olympics.txt', 'r')

# read the first line
line = olympics.readline()
print(line)

# read the second line
line = olympics.readline()
print(line)

# read all the rest of the lines
lines = olympics.readlines()
print(lines[:3]) # print the first 3 lines as an example

# close the file
olympics.close()
```

**Reading entire files**

```python
# # read the full text
# with open('data/olympics.txt', 'r') as olympics:
#     txt = olympics.read()
#     print(txt[:100])

# loop through the file reading line by line
# save medal winners to a new list
medal_winners = []

with open('data/olympics.txt', 'r') as olympics:
    # skip the header line
    olympics.readline()
    # loop through each line
    for line in olympics:
        # strip new line characters and split the line into columns
        cols = line.strip().split(',')
        # if the medal column doesn't equal 'NA', add the line to the medal winners list
        if cols[5] != 'NA':
            medal_winners.append(line.strip())

print(medal_winners[:3])
```

**Dictionaries as accumulators**

```python
# create a dictionary of grade values from csv file

gradeValues = {}
with open('data/grading_scheme.csv', 'r') as grading_scheme:
    # skip the header line
    grading_scheme.readline()
    # loop through each line
    for line in grading_scheme:
        # strip new line characters and split the line into columns
        cols = line.strip().split(',')
        # add the grade and value to the dictionary
        gradeValues[cols[0]] = float(cols[2])

print(gradeValues)
```

```python
text = 'This is a test of the emergency broadcast system. This is only a test.'

freq_words = {}
for word in text.split():
    freq_words[word] = freq_words.get(word, 0) + 1
    
    # the above is equivalent to:

    # if word not in freq_words:
    #     freq_words[word] = 0
    # freq_words[word] += 1

print(freq_words)
```

```python
for grade, GPAvalue in gradeValues.items():
    if float(GPAvalue) > 0:
        print(f'{grade} is a passing grade')
```
