# While Loops

```python
alarmTime = 5
currentTime = 1

while currentTime <= alarmTime:
    print(f'The current time is {currentTime}')
    currentTime = currentTime + 1

print("stop loop")
```

**Listener Loop**

```python
todo = ['homework', 'drink water', 'pick up milk']
task = ''
while task != 'STOP':
    task = input('Enter a task (Enter STOP to quit): ')
    todo.append(task)

print(todo)
```

```python
str1 = ''
while str1 != 'y' and str1 != 'n':
    str1 = input('Enter y or n: ')
print('good job')
```

```python
num = ''
while not num.isdigit():
    num = input("Enter a number")
num = float(num)
```

```python
#  flag variable
keepGoing = True
todo = []
while keepGoing:
    todo.append(input('Enter a task (Enter STOP to quit): '))
    if todo[-1] == 'STOP':
        del todo[-1]
        keepGoing = False
print(todo)
```

**Infinite loop**

```python
i = 0

while i != '10':
    # this will run forever since i will always be an integer, not a string
    print(i)
    i = i + 1
```

```python
packing_list = ['laptop', 'cell charger', 'toothbrush', 'book']

i = 0
while i < len(packing_list):
    # this loop will run forever since we forgot to increment i
    print(f'Remember to pack your {packing_list[i]}')
    i += 1
```

**Break, Continue**

```python
# generate an acronym using only capital letters
x = 'Seidenberg School of Computer Science and Information Systems, Dyson School of Arts and Sciences'
acronym = ''
for char in x:
    if char == ',':
        # stop looping if we find a comma
        break
    if not char.isupper():
        # skip if the character is not uppercase
        continue
    acronym = acronym + char
print(acronym)
    
```

# Advanced Functions

**Optional parameters, defaults**

```python
def increment(x, i = 1):
    # increment x by i
    # if i is not provided, increment by 1
    return x + i

print(increment(5))
print(increment(10, 2))
```

**keyword parameters** - passing values by naming the variable

```python
def orderDinner(mainCourse, appetizer=None, dessert=None):
    print(f'You ordered {mainCourse}')
    if appetizer != None:
        print(f'You ordered {appetizer} as an appetizer')
    else:
        print('no appetizer')
    if dessert != None:
        print(f'You ordered {dessert} for dessert')
    else:
        print('no dessert')
    print('Enjoy your meal!')

# orderDinner('pizza', 'salad', 'ice cream')
# orderDinner('pasta', 'veggies')
# orderDinner('sushi', dessert='baklava')
# orderDinner('soup', appetizer='bread')
orderDinner('waffles')
```

**Lambda Functions**

```python
l = lambda x: x * 2

print(l(5))
```

# Sorting

```python
listA = [0, 25, 7, 10]
print(f'ListA before sorting: {listA}')
listB = sorted(listA, reverse=True)
# listC = listA.sort()
# 
print(f'ListA: {listA}')
print(f'ListB: {listB}')
# print(f'ListC: {listC}')
```

```python
# sorting strings
list1 = ['hello', 'apple', 'dog', 'boat']
print(f'List1 before sorting: {list1}')
print(f'List1 after sorting: {sorted(list1)}')
print(f'Listed sorted backward {sorted(list1, reverse=True)}')
print(f'List1 after sorting: {list1}')
```

**Custom Sorting**

```python
# sort strings by length
print(f'List1 sorted by length: {sorted(list1, key=len)}')
```

```python
# sorting dict by value
dict1 = {'a': 5, 'b': 10, 'c': 1}
sorted_keys1 = sorted(dict1, key=lambda x: dict1[x])
print(sorted_keys1)

def number_digits(d):
    return len(str(dict1[d]))

# sort dict by value length from largest to smallest
sorted_keys2 = sorted(dict1, key=number_digits, reverse=True)
print(sorted_keys2)
```

**Secondary Sorting** - sorting tuples

```python
# sorting lists of tuples
list2 = [('a', 5, 45), 
         ('b', 10),
         ('a', 5),
         ('b', 1, 17),
         ('c', 1, -3.14)]

print(f'List2 before sorting: {list2}')
print(f'List2 after sorting: {sorted(list2)}')
```

```python
# tie breakers/seconary sort

# sort strings by length, then alphabetically
fancy_words = ['incomprehensibility', 'hippopotomonstrosesquippedaliophobia', 
               'antidisestablishmentarianism', 'supercalifragilisticexpialidocious', 'queuing', 'pizzazz']

# return a tuple with negative length (to print long words first) then the word itself to sort alphabetically
print(sorted(fancy_words, key=lambda x: (-len(x), x)))
```
