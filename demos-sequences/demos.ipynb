**String Indexes**

```python
# "Fire and Ice" by Robert Frost
a_poem = """Some say the world will end in fire,
Some say in ice.
From what I’ve tasted of desire
I hold with those who favor fire.
But if it had to perish twice,
I think I know enough of hate
To say that for destruction ice
Is also great
And would suffice."""

print('First character: ', a_poem[0])
print('Last character: ', a_poem[-1])
print('First 5 characters: ', a_poem[:5])
```

```python
a_poem[0] = 's'  # This will fail
```

**List Indexes**

```python
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'Novemeber', 'December']

print("First month: ", months[0])
print("Last month: ", months[-1])
print("11th month: ", months[10])

# oops! fix the spelling of November
months[10] = 'November'
print("11th month: ", months[10])
```

**For Loops**

```python
# count the number of 'l' characters present in the names of months
# this is a typical accumulator pattern
search_char = 'l'
l_count = 0
for month in months:
    l_count += month.count(search_char)

print(f"Number of '{search_char}' characters: {l_count}")
```

```python
# collect the every 10th characters of Fire and Ice using a for loop and range
tenth_chars = []

for i in range(0, len(a_poem), 10):
    tenth_chars.append(a_poem[i])

print(tenth_chars)
```

**Multidimensional sequences**

```python
# tic-tac-toe board
board = [['O', 'X', 'O'], ['X', 'X', 'O'], ['O', 'X', 'X']]
# X won!
for row in board:
    for col in row:
        print(col, end=' ')
    print()
```
