# Classes

**Custom Objects**

```python
class Pen:
    def __init__(self, incoming_color, incoming_category, incoming_text = ''):
        self.color = incoming_color
        self.category = incoming_category
        self.text = incoming_text
        self.lifespan = '1 yr'

    # get methods

    def get_color(self):
        return self.color
    
    def set_color(self, new_color):
        self.color = new_color
        return self.color
    
    def get_category(self):
        return self.category

    def get_text(self):
        return self.text

    def __str__(self):
        return f'{self.color} {self.category} pen'
    
    def __repr__(self) -> str:
        return self.__str__()
```

**Instantiate a new Object**

```python
# new pen
pen1 = Pen('blue', 'ballpoint', 'Rhinebeck Equine LLP')

# view things about pen1
print(pen1) # call __str__
print(f'Pen1 color: {pen1.get_color()}')
print(f'Pen1 category: {pen1.get_category()}')
print(f'Pen1 text: {pen1.get_text()}')
pen1.set_color('orange')
print(pen1)
print(pen1.lifespan)
print()


# # another pen
pen2 = Pen('black', 'felt tip')

# view things about pen2
print(pen2)
print(f'Pen2 color: {pen2.get_color()}')
print(f'Pen2 category: {pen2.get_category()}')
print(f'Pen2 text: {pen2.get_text()}')

pens = [pen1, pen2]
print(pens)
```

**str**

```python
class Obj:
    def __init__(self, name):
        self.name = name

obj1 = Obj('key')
print(f'First object: {obj1}')

class Obj_w_str:
    def __init__(self, name):
        self.name = name
    def __str__(self) -> str:
        return self.name
    def __repr__(self) -> str:
        return self.__str__()
    
obj2 = Obj_w_str('another key')
print(f'Second object: {obj2}')
```

**self**

```python
name = 'Will'
greeting = 'Boo'

class Hello:
    greeting = 'Hi!'

    def __init__(self, incoming_name):
        self.name = incoming_name
    
    def whoami(self):
        print(self.name)
        # print(name) # this prints Will every time
    
    def sayhi(self):
        print(self.greeting)
        # print(greeting) # this prints Boo every time

    def get_globals(self):
        print(f'{greeting} {name}')

me = Hello('Leanne')
me.whoami()
me.sayhi()
print()

you = Hello('Julian')
you.whoami()

you.get_globals()
```

## Program Structure

```python
# import statements
import random

# declare classes
class Flashlight:
    settings = (
        'off',
        'red',
        'dim',
        'bright'
    )
    def __init__(self, lumens):
        self.lumens = lumens
        self.state = 0
    def get_state(self):
        return self.settings[self.state]
    def get_lumens(self):
        return self.lumens
    def click(self):
        self.state = (self.state + 1) % len(self.settings)
        return self.get_state()
    def __str__(self):
        return f'Flashlight is {self.get_state()}'
    
# declare functions
def turn_off_flashlight(flashlight):
    while flashlight.get_state() != 'off':
        flashlight.click()
    print('Flashlight is off')

# driver code
headlamp = Flashlight(100)
headlamp.click()
print(headlamp)
headlamp.click()
print(headlamp)
turn_off_flashlight(headlamp)
print(headlamp.get_lumens())
```
