**Logical Operators**

```python
print(f"True and True -> {(True and True)}")
print(f"True and False -> {(True and False)}")
print(f"False and True -> {(False and True)}")
print(f"False and False -> {(False and False)}")
print()

print(f"True or True -> {(True or True)}")
print(f"True or False -> {(True or False)}")
print(f"False or True -> {(False or True)}")
print(f"False or False -> {(False or False)}")
print()

print(f"not True -> {(not True)}")
print(f"not False -> {(not False)}")
```

**Transforming Lists, Tuples, Strings**

```python
fruitList = ["apple", "banana", "orange", "pear", "grape", "pineapple"]

# change pineapple to mango
fruitList[5] = "mango"
print(fruitList)

fruitTuple = ("apple", "banana", "orange", "pear", "grape", "pineapple")
# change pineapple to mango
#fruitTuple[5] = "mango" # ERROR
fruitTuple = fruitTuple[:5] + ("mango",) # OK
print(fruitTuple)

fruitString = "apple, banana, orange, pear, grape, pineapple"
# change pineapple to mango
#fruitString[36:] = "mango" # ERROR
fruitString = fruitString[:36] + "mango" # OK
print(fruitString)
```

**String methods**
<br>
nonmutating

```python
testString = "Hello World"

print(f"Upper: {testString.upper()}")
print(f"Lower: {testString.lower()}")
print(f"Count of o: {testString.count('o')}")
print(f"Index of o: {testString.index('o')}")
print(f"Replace o with 0: {testString.replace('o', 'i')}")
print()
print(f"Original string is unchanged: {testString}")
```

```python
testString = "Hello World"
num = testString.count('o')
indexes = []
lastIndex = 0
for _ in range(num):
    indexes.append(testString.index('o', lastIndex))
    lastIndex = indexes[-1] + 1 

print(indexes)
```

**List methods**
<br>
mutating and nonmutating

```python
testList = ["Maple", "Oak", "Pine", "Birch", "Elm", "Ash"]
print(f"Original list: {testList}")
testList.append("Cedar")
# print(testList.append("Cedar"))
print(f"Append to list: {testList}")
testList.insert(2, "Willow")
print(f"Insert into list: {testList}")
print(f"Pop 4th item from list: {testList.pop(3)}\nResulting list: {testList}")
testList.sort()
print(f"Sort list: {testList}")
testList.reverse()
print(f"Reverse list: {testList}")
print(f"Index of Elm: {testList.index('Elm')}")
print(f"Count of Redwood: {testList.count('Redwood')}")
testList.remove("Elm")
print(f"Remove Elm from list: {testList}")
```

**Aliasing**
<br>
Lists and Strings

```python
# aliasing lists
a = [1, 2, 3]
b = a
b[0] = 5 # this changes both a and b
print(f"a: {a}")
print(f"b: {b}")
print()

# aliasing strings
a = "Hello"
b = a
a = "Goodbye" # this does not change b
print(f"a: {a}")
print(f"b: {b}")
```

```python
# string formatting examples
# set to current date for best results
day = 8
month = 10
year = 2024



print('Today is {}-{}-{}'.format(day, month, year))
print("Today is {1}/{0}/{2}".format(day, month, year-2000))
print(f'Today is the {day}th day of the {month}th month of the year {year}')
# print('Today is', day, month, year)
# this is another old way of formatting strings
print("Today is %d/%d/%d" % (day, month, year))
```
