## Function Scope

**Local vs Global variables**

```python
global_var = 10

def scope_test():
    print('entering function')
    local_var = 20
    print(f'local var: {local_var}')
    print(f'global var: {global_var}')
    print('exiting function')


scope_test()
print('outside function')
print(f'global var: {global_var}') 
print(f'local var: {local_var}')  # this causes an error
```

**Passing mutable vs immutable function parameters**

```python
immutable = 10
mutable = [1, 2, 3]

def test_function():
    immutable = 20
    mutable.append(4)
    print(f'inside function immutable: {immutable}')
    print(f'inside function mutable: {mutable}')
    print()

print(f'before function immutable: {immutable}')
print(f'before function mutable: {mutable}')
print()
test_function()
print(f'outside function immutable: {immutable}')
print(f'outside function mutable: {mutable}')
```

## Tuple Packing

```python
# packing values into a tuple

valid_tuple = 1, 2, 3
```

```python
# unpacking values from a tuple

a, b, c = valid_tuple
```

```python
# using tuple packing to return multiple values from a function

def min_max(numbers):
    smallest = min(numbers)
    largest = max(numbers)
    return smallest, largest

least, greatest = min_max([3,4,5,6,7,8,9,0])
print(f'least: {least}, greatest: {greatest}')
```

```python
# using tuple unpacking with the enumerate function
my_list = ['a', 'b', 'c', 'd', 'e']

for index, value in enumerate(my_list):
    print(f'index: {index}, value: {value}')

# for i in range(len(my_list)):
#     print(f'index: {i}, value: {my_list[i]}')
```

```python
fruits = {
    'apple': 'red',
    'pear': 'green',
    'tomato': 'yellow',
    'orange': 'orange',
    'grape': 'purple'
}

for fruit, color in fruits.items():
    print(f'{fruit} is {color}')
```
